```python
import pandas as pd

# Point directly to the Excel file inside the subfolder
file_path = r"C:\Users\Nathan\Downloads\Supply_Chain_Analysis_of_Inventories_Data_Analytics\Supply_Chain_Analysis_of_Inventories_Data_Analytics_with_Power_B.xlsx"

# Read your sheets
sales_df = pd.read_excel(file_path, sheet_name="Sales")
product_df = pd.read_excel(file_path, sheet_name="Product")
customer_df = pd.read_excel(file_path, sheet_name="Customer")

print("Files loaded successfully!")
```

```python
print("Sales duplicates:", sales_df.duplicated().sum())
print("Product duplicates:", product_df.duplicated().sum())
print("Customer duplicates:", customer_df.duplicated().sum())
```

```python
sales_df['Calculated_Total'] = (sales_df['Quantity'] * sales_df['UnitPrice']) - sales_df['Discount']
# Check if there is any mismatch between given TotalAmount and calculated total
mismatches = sales_df[abs(sales_df['TotalAmount'] - sales_df['Calculated_Total']) > 0.01]
print("Mismatched records:", len(mismatches))
```

```python
# 1. Drop duplicate rows
sales_df = sales_df.drop_duplicates()

# 2. Drop the temporary helper column
if 'Calculated_Total' in sales_df.columns:
    sales_df = sales_df.drop(columns=['Calculated_Total'])

# 3. Export the clean CSVs
sales_df.to_csv("cleaned_sales.csv", index=False)
product_df.to_csv("cleaned_products.csv", index=False)
customer_df.to_csv("cleaned_customers.csv", index=False)

print("Cleaned CSV files saved successfully!")
```
